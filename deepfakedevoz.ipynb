```python
import torch
from TTS.api import TTS
#primeiro projeto realizado em um macbook
#desenvolvido por Jhonn Vyctor
print("desenvolvido por Jhonn Vyctor")
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Dispositivo:", device)

# Baixa/carrega o XTTS-v2 (na primeira execução pode demorar)
modelo = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)
print("XTTS-v2 carregado.")
```

## Envie a gravação de referência

Prefira uma gravação limpa, sem música ou ruído, com uma única pessoa falando. O áudio será usado como referência vocal.

```python
from google.colab import files

enviados = files.upload()
arquivo_referencia = next(iter(enviados))
print("Áudio de referência:", arquivo_referencia)
```

## Escreva o texto em português

O código de idioma do XTTS-v2 para português é `pt` (não `pt-BR`).

```python
texto = input("escreva o texto a ser dito.")
arquivo_saida = input("coloque o nome do audio por favor e coloque a extensao do arquivo(mp3,wav)")
modelo.tts_to_file(
    text=texto,
    speaker_wav=arquivo_referencia,
    language="pt",
    file_path=arquivo_saida,
    split_sentences=True
)

print("Áudio gerado:", arquivo_saida)
```

Conserto de audio(opcional, esse trecho pode ser pulado)


```python
import soundfile as sf
import numpy as np

audio,taxa = sf.read(arquivo_saida, dtype="float32")
audio = np.nan_to_num(audio)
audio = audio - np.mean(audio, axis=0)
pico = np.max(np.abs(audio))
if pico >0:
  ganho_maximo = 10 ** (3/10)
  ganho = min(0.95 / pico, ganho_maximo)
  audio = audio  * ganho
sf.write(arquivo_saida , audio, taxa)
print("áudio ajustado")
```
